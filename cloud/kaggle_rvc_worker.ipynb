# AI Vocals Studio: Free Kaggle GPU RVC Worker

1. In Kaggle, enable **Settings -> Accelerator -> GPU**.
2. Add Kaggle Secrets named `SUPABASE_URL`, `SUPABASE_SERVICE_ROLE_KEY`, and `SUPABASE_STORAGE_BUCKET` (`voiceovers`).
3. Run all cells. The notebook claims one queued `Train and register RVC model` job, trains with Applio, and uploads the model/index to Supabase.

Free GPU availability and quotas are controlled by Kaggle. Stop the session when finished.

```python
import os, subprocess, sys
from pathlib import Path

APP_ROOT = Path('/kaggle/working/ai-vocals-studio')
if not (APP_ROOT / '.git').exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/coden607/ai-vocals-studio.git', str(APP_ROOT)], check=True)
APPLIO_ROOT = Path('/kaggle/working/Applio')
if not (APPLIO_ROOT / '.git').exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/IAHispano/Applio.git', str(APPLIO_ROOT)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(APPLIO_ROOT / 'requirements.txt')], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'requests', 'librosa', 'soundfile'], check=True)
os.environ['APP_ROOT'] = str(APP_ROOT)
os.environ['APPLIO_ROOT'] = str(APPLIO_ROOT)
subprocess.run([sys.executable, str(APP_ROOT / 'cloud' / 'kaggle_rvc_worker.py')], check=True)
```
